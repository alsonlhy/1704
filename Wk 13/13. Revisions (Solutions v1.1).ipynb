### 6.

Write a program that takes in a string and prints out the characters with odd indices.


```
Enter a sentence :Blue Lagoon
position 1: l
position 3: e
position 5: L
position 7: g
position 9: o

total: 5 characters at odd positions
```


````python
# Solution

sentence = input('Enter a sentence :')
count = 0
for i in range(len(sentence)):
    if i % 2 == 1:
        ch = sentence[i]
        print('position ' + str(i) + ' : ' + ch)
        count += 1

print('\nThe total number of characters at odd positions is', count)
````

### 7.

Write a program that prompts the user for a string and displays the string where all the vowels (both uppercase and lowercase) ('a', 'e', 'i', 'o', 'u') have been replaced by *.

```
Enter a sentence :Mississipi River
New sentence:M*ss*ss*p* R*v*r
```

````python
# Solution

sentence = input('Enter a sentence :')

vowel = 'aeiou'
final_str = ''
for ch in sentence:
    ch_lower = ch.lower()
    if ch_lower in vowel:
        final_str += '*'
    else:
        final_str += ch

print('Here is the sentence after hiding the vowels :' + final_str)
````

E.g. ```get_total_length(['123', '00', 'abcde'])``` should return 10.


````python
# Solution 1

def get_total_length(my_list):
    total_str = ''
    for s in my_list:
        total_str += s
    return len(total_str)
````

### 12. Keep prompting the user for input until a criteria is fulfilled.

Write some code that prompts the user for two integer numbers num1 and num2 and stops prompting the user when

(1) num1 is a multiple of 3 and num2 is a multiple of 5,

__or__

(2) when the sum of the two numbers is a multiple of 7.

```
first number : 4
second number : 5
Not valid!

first number : 6
second number : 6
Not valid!

first number : 6
second number : 10
Bye!
```
other sample run
```
first number : 17
second number : 4
Bye!
```

````python
# Solution - 1

num1 = int(input("Please enter a first number : "))
num2 = int(input("Please enter a second number : "))

while not( (num1 % 3 == 0 and num2 % 5 == 0) or (num1+num2) % 7 == 0 ):
    print("Not valid!")
    num1 = int(input("\nPlease enter a first number : "))
    num2 = int(input("Please enter a second number : "))

print('Bye!')
````

````python
# Solution - 2

num1 = int(input("Please enter a first number : "))
num2 = int(input("Please enter a second number : "))

while (num1 % 3 != 0 or num2 % 5 != 0) and (num1+num2) % 7 != 0:   # and has higher precedence than or
    print("Not valid!")
    num1 = int(input("\nPlease enter a first number : "))
    num2 = int(input("Please enter a second number : "))

print('Bye!')
````

### 14. reverse segments

Define a function called ```reverse_segments()``` that takes in a string called ```text``` as its parameter.

```text``` contains segments of texts separated by one of the following characters: ```|-=```.

For example, ```text``` may be ```'a b c|def-123 aaa=|(abc)'``` .

The segments inside the string are:

```
'a b c' ,  'def' , '123 aaa' , '' , '(abc)'
```

The function segments the input string and reverses each segment character by character. It then returns all the reversed segments as a list.

For example, given the input string above, the function returns the following list:

```
['c b a', 'fed', 'aaa 321', '', ')cba(']
```

````python
# solution 1
def reverse_segments(text):
    segment = ''
    result_list = []
    for ch in text:
        if ch in '|-=':
            result_list.append(segment[::-1])
            segment = ''
        else:
            segment += ch

    result_list.append(segment[::-1])
    return result_list

# solution 2:
def reverse_segments(text):
    return_text = []
    text = text.replace('-', '|')
    text = text.replace('=', '|')
    list_s = text.split('|')
    for s in list_s:
        return_text.append(s[::-1])
    return return_text

print(reverse_segments('a b c|def-123 aaa=|(abc)'))
````
