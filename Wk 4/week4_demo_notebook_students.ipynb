### 8. for loops to iterate through all the characters of a string

#### Two ways to iterate through the characters of a string

```python
s = 'abcdef'
```

```python
# Method 1
#---------

for ch in s:
    print(ch, end=' ')
```

```python
# Method 2
#---------

for i in range(0, len(s)):
    ch = s[i]
    print(ch, end=' ')    
```

#### When do we need method 2?

```python
# example 1
s = 'abcdef'
for i in range(len(s)):
    print('Index', i, ':', s[i])    
```

```python
# example 2
s = 'abcdef'
for i in range(0,len(s)):
    if i % 2 == 0:
        print(s[i], end=' ')    
```

#### Index of the last character?

```python
print(s[len(s) - 1])
```

```python
print(s[-1])
```
