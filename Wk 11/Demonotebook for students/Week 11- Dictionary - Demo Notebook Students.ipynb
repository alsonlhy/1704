## <font color = 'blue'>Efficiency </font>

```python
# ----------
# Example 1
# ----------

a_dict = {"John": "one", "Mary": "two", "Bob": "three"}

name = input("Enter a name: ")
if name in a_dict:
    for key in a_dict:
        if key == name:
            print(a_dict[name])
else:
    print("The name is not in the dictionary")
```

What is the efficiency of this code?

```python
# Write your solution here


```

Solution:

- `name in a_dict` → O(1) on average.
- `for key in a_dict` → O(n) in the worst case, because we may need to examine all the keys.
- `key == name` → O(1), assuming constant-time string comparisons.
- `a_dict[name]` → O(1) on average.

Therefore, the overall worst-case time complexity is <b>O(n) (Linear time)</b>.

Although checking whether a key exists in a dictionary takes O(1) time on average, the for loop searches through the keys one by one. <br>In the worst case, it may examine all n keys.

This is an <b><font color = 'red'>inefficient way to use a dictionary!</font></b> <br>We are not taking advantage of its ability to access values directly using their keys.

```python
# --------------------------------------
# Example 2 : Better usage of dictionary
# --------------------------------------

a_dict = {"John": "one", "Mary": "two", "Bob": "fourth"}

name = input("Enter a name: ")
if name in a_dict:
    print(a_dict[name])
else:
    print("The name is not in the dictionary")
```

What is the efficiency of this code?

```python
# Write your solution here:



```

Solution:

- Time complexity: <b>O(1) (Constant time on average)</b>

- By using `name in a_dict` and `a_dict[name]`, we can check whether a key exists and retrieve its associated value <b>directly</b>, without iterating through the dictionary.

The average time complexity remains O(1), <u>regardless of the number of entries in the dictionary</u>.<br>
This is an <b><font color = 'green'>efficient way to use a dictionary!</font></b> 

## <font color = 'blue'>How does a Python dictionary work (internally)?</font>

### 1. Python dictionaries use a technique called <b>hashing</b> to store and retrieve key-value pairs efficiently.<br>
Let's start with a dictionary containing two entries.

```python
phone_book = {
    "Lawrence Wong": 67428832,
    "Goh Chok Tong": 69226844
}
```

### 2. Inserting a new key-value pair
Suppose we want to add a new entry:

```python
phone_book["Lee Hsien Loong"] = 95556411
```

What happens internally?

1. Python applies a <b>hash function</b> to the key `"Lee Hsien Loong"` to calculate a <b>hash value</b> (an integer).
2. Python uses this <b>hash value</b> to determine a <b>location (slot) in its internal hash table</b>.
3. Python checks whether the key already exists:
   - If the key does not exist, the new key-value pair is inserted.
   - If the key already exists, its associated value is updated.

We can see the <b>hash value</b> using Python's built-in `hash()` function:

```python
hash("Lee Hsieng Loong")
```

Note: 
- The <b>hash value</b> for a string may differ between Python sessions.
- Python does not need to search through all existing keys one by one to insert or update an entry.

Therefore, <b>inserting</b> or <b>updating</b> a dictionary entry takes <b>O(1) time</b> on average.

### 3. Retrieving a value using its key

```python
print(phone_book["Lee Hsien Loong"])
```

What happens internally?

1. Python applies the same <b>hash function</b> to the key `"Lee Hsien Loong"`.
2. Python uses the <b>hash value</b> to determine where to look in its internal hash table.
3. Python <b>locates the entry</b> associated with the key and retrieves its value.

### 4. Why Time Complexity is O(1)?

With a <b>dictionary</b>, Python uses <b>hashing</b> to locate the relevant entry <b>directly</b>, without searching sequentially through all keys.

This is why dictionary operations such as <b>insertion</b>, <b>updating</b>, and <b>retrieval</b> take <b>O(1) time</b> on average, even when the dictionary contains many entries.

## Summary:

- Checking for key in dictionary -> O(1)
- Setting or getting a value in a dictionary -> O(1)
- Getting all the keys of a dictionary -> O(1)
- Getting all the values of a dictionary -> O(1)
- Getting all the key/value pairs of a dictionary -> O(1)


## <font color = 'blue'>Dictionary Use Cases</font>

### As Counter (to record frequency of some event)

```python
# Read the file and count word frequency using a dictionary

filename = "sample_text.txt"
word_freq = {}

with open(filename, "r") as file:
    for line in file:
        words = line.strip('\n').lower().split(' ')
        for word in words:
            # Remove punctuation from each word
            word = word.strip(".,!?;:")
            if word not in word_freq:
                word_freq[word] = 1
            else:
                word_freq[word] += 1
                # word_freq[word] = word_freq[word] + 1

# Display the word frequencies
for word, count in word_freq.items():
    print(f"{word}: {count}")
```

### As Storage of List of related item

A dictionary in Python can be used not only to <b>count the occurrence of events</b> (such as word frequencies) but also to <b>store collections of related items</b>, <br>
for example, keeping a list of phone numbers under the same person’s name.

```python
phonebook_dict = {}
with open('phone_numbers.txt', 'r') as my_file:
    for line in my_file:
        line = line.rstrip('\n')
        person, number = line.split('|')
        if person not in phonebook_dict:
            phonebook_dict[person] = [ number ]
        else:
            existing_numbers = phonebook_dict[person]
            existing_numbers.append(number)
# print(phonebook_dict)

for name, numbers in phonebook_dict.items():
    print(name)
    for number in numbers:
        print(number)
    print()
```

Much more efficient than with Lists!!!
