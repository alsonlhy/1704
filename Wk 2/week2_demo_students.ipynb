## 2. Function Call

ONLY after a function is defined, you can start calling that function

### A. Docstring

````python
# Written by the person who writes the function 
# to describe what to expect from the function

def print_sum(num1, num2):
    """
    
    """
    print(num1 + num2)
````

````python
# Written by the person who writes the function

def print_sum(num1, num2):
    """
    The function calculates the sum of the two numbers num1 and num2.
    
    Parameters: 
        num1: the first number to add
        num2: the second number to add
        
    Returns:
        The sum of num1 and num2.  
    
    """
    print(num1 + num2)
````

### C. Return Statement

You MAY or NOT have a `return` statement

````python
def calculate_sum(num1, num2):
    num_sum = num1 + num2
    return num_sum
````

````python
my_sum = calculate_sum(3.5, 8)
````

````python
print('The sum of the 2 numbers is: ' + str(my_sum))
print('The sum of the 2 numbers is:' , my_sum)
print(f'The sum of the 2 numbers is: {my_sum}')
````

### D. Flow of Execution

````python
# MSD

def add_numbers(a, b):
    a = a + 1
    b = b * 2
    return a + b

a = 1
b = 3
c = add_numbers(a, b)

print(a)
print(b)
print(c)
````
