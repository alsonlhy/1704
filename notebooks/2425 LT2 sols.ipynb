```python
# Name:
# Email ID:


def get_swim_report(filename, swimmer_name, event_name):
    # Replace the code below with your implementation.
    
    tot_time = 0
    num_events = 0
    with open(filename, 'r') as my_file:
        for line in my_file:
            line = line.rstrip('\n')
            swimmer, event, time_str = line .split(',')
            if swimmer == swimmer_name and event == event_name:
                tot_time += int(time_str)
                num_events += 1
    if tot_time == 0:
        return (0, 0)

    avg = round(tot_time / num_events)
    return (num_events, avg)
                
```

```python
# Name:
# Email ID:

def calculate_profit_margin(selling_price, buying_price):
    return (selling_price - buying_price) / buying_price


def calculate_min_stock(profit_margin):
    if profit_margin >= 0.5:
        return 3
    elif 0.2 <= profit_margin < 0.5:
        return 8
    else:
        return 15
    
def clean_price(price_str):
    if price_str[0] == '$':
        return float(price_str[1:])
    else:
        return float(price_str)
    

def manage_inventory(filename):
    # Replace the code below with your implementation.
    
    inventory_dict = {}
    with open(filename, 'r') as my_file:
        for line in my_file:
            line = line.strip('\n')
            product, category, selling_price_str, buying_price_str, qty_str = line.split(',')
            selling_price = clean_price(selling_price_str)
            buying_price = clean_price(buying_price_str)
            qty = int(qty_str)
            profit_margin = calculate_profit_margin(selling_price, buying_price)
            min_stock = calculate_min_stock(profit_margin)
            inventory_dict[product] = [qty, min_stock]
    
    return inventory_dict
```

```python
# Name:
# Email ID:

def process_orders(filename, list_of_orders):
    order_dict = {}
    
    inventory_dict = manage_inventory(filename)
    for order in list_of_orders:
        qty_filled = 0
        order_no, buyer, product, qty_ordered = order
        if product in inventory_dict:
            qty_in_stock = inventory_dict[product][0]
            if qty_in_stock >= qty_ordered:
                qty_filled = qty_ordered
                # update of the stock
                inventory_dict[product][0] -= qty_ordered
            else:
                qty_filled = qty_in_stock
                # update of the stock
                inventory_dict[product][0] = 0
            
        # update of order_dict
        if buyer not in order_dict:
            order_dict[buyer] = [ (order_no, product, qty_ordered, qty_filled) ]
        else:
            order_dict[buyer].append( (order_no, product, qty_ordered, qty_filled) )
    
    return order_dict
```

```python
# Name:
# Email ID:
        
def get_invalid_transactions(filename):
    invalid_transactions = []
    miner_unspent_dict = {}
    with open(filename, 'r') as my_file:
        for line in my_file:
            line = line.rstrip('\n')
            line_list = line.split('|')
            trans_id = line_list[0]
            
            if 'COINBASE' in line:
                out = line_list[2]
                out_list = out.split('Miner')
                # out_list is ['OUT:'', 'A:9'] for example
                out_data_str = out_list[1]
                # out_data_str is e.g. 'A:9'
                final_list = out_data_str.split(':') 
                miner = 'Miner' + final_list[0]
                # e.g. 'MinerA'
                value = int(final_list[1])
                if miner not in miner_unspent_dict:
                    # we add 1 coin in a list
                    miner_unspent_dict[miner] = {value:1}
                else:
                    # we add another coin in the list of that miner
                    miner_value_dict = miner_unspent_dict[miner]
                    if value not in miner_value_dict:
                        miner_value_dict[value] = 1
                    else:
                        miner_value_dict[value] += 1           
            else:
                line_list = line_list[1:]
                
                if not is_valid_transaction(miner_unspent_dict, line_list):
                    # transaction is invalid
                    invalid_transactions.append(trans_id)
                    
#     #clean miner_unspent_dict
#     final_miner_unspent_dict = {}
#     for key in miner_unspent_dict:
#         if miner_unspent_dict[key] != {}:
#             final_miner_unspent_dict[key] = miner_unspent_dict[key]
            
#     miner_unspent_dict = final_miner_unspent_dict
            
    #print(final_miner_unspent_dict)
    return invalid_transactions

def is_valid_transaction(miner_unspent_dict, line_list):
    
    tot_in = 0
    tot_out = 0
    transaction_in_dict = {}
    transaction_out_dict = {}
    
    #line_list is e.g. ['IN:Alice:8','IN:Bob:10','OUT:David:12','OUT:Alice:1','OUT:Charlie:5']
    for item in line_list:
        item_list = item.split(':')
        miner = item_list[1]
        amount = int(item_list[2])
        if 'IN' in item:
            tot_in += amount
            if miner not in transaction_in_dict:
                transaction_in_dict[miner] = {amount:1}
            else:
                # we add another coin to spend for that miner
                miner_amount_in_dict = transaction_in_dict[miner]
                if amount not in miner_amount_in_dict:
                    miner_amount_in_dict[amount] = 1
                else:
                    miner_amount_in_dict[amount] += 1    
                    # transaction_in_dict is e.g. {'Alice':{11:2, 4:1}, 'Bob':{10:1}}   
        if 'OUT' in item:
            tot_out += amount
            if miner not in transaction_out_dict:
                transaction_out_dict[miner] = {amount:1}
            else:
                # we add another coin to spend for that miner
                miner_amount_out_dict = transaction_out_dict[miner]
                if amount not in miner_amount_out_dict:
                    miner_amount_out_dict[amount] = 1
                else:
                    miner_amount_out_dict[amount] += 1    
                    # transaction_out_dict is same format as transaction_in_dict
    
    # verification 1 : all amount IN must be equal to all amount OUT
    if tot_in != tot_out:
        return False
    
    # verification 2 : 
    # each amount in transaction_in_dict for a particular miner must exist in unspent_dict
    for miner in transaction_in_dict:
        # transaction_in_dict is e.g. {'Alice':{11:2, 4:1}, 'Bob':{10:1}} 
        # unspent_dict is e.g. {'Alice':{11:1, 4:1}, 'Bob':{10:1}, 'Charlie':{5:1}} 
        if miner not in miner_unspent_dict:
            return False
        
        # miner is in both dictionaries: transaction_in_dict and miner_unspent_dict
        miner_amount_in_dict = transaction_in_dict[miner]
        miner_amount_unspent_dict = miner_unspent_dict[miner]
        
        for amount in miner_amount_in_dict:
            if amount not in miner_amount_unspent_dict:
                # miner cannot spend an amount that is not in miner_unspent_dict
                return False
            
            # let verify the quantity now (number of coins)
            num_coins_to_be_spent = miner_amount_in_dict[amount]
            num_coins_that_can_be_spent = miner_amount_unspent_dict[amount]
            if num_coins_to_be_spent > num_coins_that_can_be_spent:
                return False   
       
    # No rule is violated, so the transaction is valid
    # let's update miner_unspent_dict to reflect the transaction
    update_valid_transaction(miner_unspent_dict, transaction_in_dict, transaction_out_dict) 
    return True
  
    
def update_valid_transaction(miner_unspent_dict, transaction_in_dict, transaction_out_dict):
    
    # remove the coin corresponding to the amount spent by the miner
    for miner in transaction_in_dict:
        #print('\tminer, amount:', miner, amount)
        miner_amount_in_dict = transaction_in_dict[miner]
        for amount in miner_amount_in_dict:
            num_coins = miner_amount_in_dict[amount]
            # remove this from miner_unspent_dict
            miner_amount_unspent_dict = miner_unspent_dict[miner]
            miner_amount_unspent_dict[amount] -= num_coins
            if miner_amount_unspent_dict[amount] == 0:
                # remove that amount from the dictionary
                miner_amount_unspent_dict.pop(amount)
                
    # add new coin(s) to the addressee corresponding to the amount received
    for miner in transaction_out_dict:
        miner_amount_out_dict = transaction_out_dict[miner]
        
        for amount in miner_amount_out_dict:
            num_coins = miner_amount_out_dict[amount]
            # add this from miner_unspent_dict
            if miner not in miner_unspent_dict:
                miner_unspent_dict[miner] = {amount:num_coins}
            else:
                miner_amount_unspent_dict = miner_unspent_dict[miner]
                if amount not in miner_amount_unspent_dict:
                    miner_amount_unspent_dict[amount] = num_coins
                else:
                    miner_amount_unspent_dict[amount] += num_coins
                    
    #clean miner_unspent_dict
    final_miner_unspent_dict = {}
    for key in miner_unspent_dict:
        if miner_unspent_dict[key] != {}:
            final_miner_unspent_dict[key] = miner_unspent_dict[key]
            
    miner_unspent_dict = final_miner_unspent_dict
    #print(miner_unspent_dict)               
```

```python
def build_valid_transaction(miner_unspent_dict, line_list):
   
    transaction_in_dict = {}
    transaction_out_dict = {}
    
    #line_list is e.g. ['IN:Alice:8','IN:Bob:10','OUT:David:12','OUT:Alice:1','OUT:Charlie:5']
    for item in line_list:
        item_list = item.split(':')
        miner = item_list[1]
        amount = int(item_list[2])
        if 'IN' in item:
            if miner not in transaction_in_dict:
                transaction_in_dict[miner] = {amount:1}
            else:
                # we add another coin to spend for that miner
                miner_amount_in_dict = transaction_in_dict[miner]
                if amount not in miner_amount_in_dict:
                    miner_amount_in_dict[amount] = 1
                else:
                    miner_amount_in_dict[amount] += 1    
                    # transaction_in_dict is e.g. {'Alice':{11:2, 4:1}, 'Bob':{10:1}}   
        if 'OUT' in item:
            if miner not in transaction_out_dict:
                transaction_out_dict[miner] = {amount:1}
            else:
                # we add another coin to spend for that miner
                miner_amount_out_dict = transaction_out_dict[miner]
                if amount not in miner_amount_out_dict:
                    miner_amount_out_dict[amount] = 1
                else:
                    miner_amount_out_dict[amount] += 1    
                    # transaction_out_dict is same format as transaction_in_dict
    
    # remove the coin corresponding to the amount spent by the miner
    for miner in transaction_in_dict:
        #print('\tminer, amount:', miner, amount)
        miner_amount_in_dict = transaction_in_dict[miner]
        for amount in miner_amount_in_dict:
            num_coins = miner_amount_in_dict[amount]
            # remove this from miner_unspent_dict
            miner_amount_unspent_dict = miner_unspent_dict[miner]
            miner_amount_unspent_dict[amount] -= num_coins
            if miner_amount_unspent_dict[amount] == 0:
                # remove that amount from the dictionary
                miner_amount_unspent_dict.pop(amount)
                
    # add new coin(s) to the addressee corresponding to the amount received
    for miner in transaction_out_dict:
        miner_amount_out_dict = transaction_out_dict[miner]
        
        for amount in miner_amount_out_dict:
            num_coins = miner_amount_out_dict[amount]
            # add this from miner_unspent_dict
            if miner not in miner_unspent_dict:
                miner_unspent_dict[miner] = {amount:num_coins}
            else:
                miner_amount_unspent_dict = miner_unspent_dict[miner]
                if amount not in miner_amount_unspent_dict:
                    miner_amount_unspent_dict[amount] = num_coins
                else:
                    miner_amount_unspent_dict[amount] += num_coins
       
    #clean miner_unspent_dict
    final_miner_unspent_dict = {}
    for key in miner_unspent_dict:
        if miner_unspent_dict[key] != {}:
            final_miner_unspent_dict[key] = miner_unspent_dict[key]
    
    return final_miner_unspent_dict
    
        
def get_miner_unspent_dict(filename):
    miner_unspent_dict = {}
    with open(filename, 'r') as my_file:
        for line in my_file:
            line = line.rstrip('\n')
            line_list = line.split('|')
            trans_id = line_list[0]
            
            # COINBASE transactions
            if 'COINBASE' in line:
                out = line_list[2]
                out_list = out.split('Miner')
                # out_list is ['OUT:'', 'A:9'] for example
                out_data_str = out_list[1]
                # out_data_str is e.g. 'A:9'
                final_list = out_data_str.split(':') 
                miner = 'Miner' + final_list[0]
                # e.g. 'MinerA'
                value = int(final_list[1])
                if miner not in miner_unspent_dict:
                    # we add 1 coin in a list
                    miner_unspent_dict[miner] = {value:1}
                else:
                    # we add another coin in the list of that miner
                    miner_value_dict = miner_unspent_dict[miner]
                    if value not in miner_value_dict:
                        miner_value_dict[value] = 1
                    else:
                        miner_value_dict[value] += 1           
            else:
                # Non COINBASE transactions
                line_list = line_list[1:]
                miner_unspent_dict = build_valid_transaction(miner_unspent_dict, line_list)
    
    return miner_unspent_dict
```

```python
from itertools import combinations

def get_coins(filename, miner, target):
    miner_unspent_dict = get_miner_unspent_dict(filename)
    # print(miner_unspent_dict)
    miner_coins_dict = miner_unspent_dict[miner]
    miner_coins_list = []
    for coin in miner_coins_dict:
        num_coins = miner_coins_dict[coin]
        for i in range(num_coins):
            miner_coins_list.append(coin)
    
    combinations_list = []
    coins_list = []
    
    for size in range(1, len(miner_coins_list)+1):
        for combination in combinations(miner_coins_list, size):
            if sum(combination) >= target and combination not in combinations_list:
                combinations_list.append(combination)

    if combinations_list == []:
        return []

    min_change = sum(combinations_list[0])
    for combination in combinations_list:
        sum_comb = sum(combination)
        if sum_comb < min_change:
            min_change = sum_comb

    # the minimum change is min_change. let's get all the combinations that provide that min_change
    for combination in combinations_list:
        coins_sum = sum(combination)
        if coins_sum == min_change:
            coins_list.append(combination)
            
    return coins_list
```
