### Adding an element in a list

```python
a_list = [3, 4, 5]
a_list.append(8)
print(a_list)
```

```python
a_list = [3, 4, 5]
a_list.append([1, 2])
print(a_list)
```

```python
print(a_list[3][1])
```

```python
print(a_list)
```

```python
# Wrong - Dont do that!
my_list = a_list.append(8)
print(my_list)
```

```python
# We can append only one element at a time!!!

a_list = [3, 4, 5]
a_list.append(88)
a_list.append(99)
a_list.append(77)
print(a_list)
```

### Iterating through a string

```python
a_str = 'abcd'
for ch in a_str:
    print(ch, end = ' ') 
```

### Iterating through a list

```python
a_list = ['a', 'b', 'c', 'd']
for ch in a_list:
    print(ch, end = ' ')
```

```python
a_list = [('a',1), ('b',5), ('c',10), ('d',7)]
for tup in a_list:
    print(tup, end = ' ')
```

```python
a_list = [('a',1), ('b',5), ('c',10), ('d',7)]
print(a_list[1][0])
```

### Iterating through a string with index

```python
a_str = 'abcd'
for i in range(len(a_str)):
    ch = a_str[i]
    print(i, '->', ch)
```
