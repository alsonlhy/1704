### <font color = 'blue'>Using boolean variables in if / else statement</font>

a) The 2 statements are independant from each other.

```python
n = int(input('Enter a number: '))

if n < 5:
    print('A')
    
if n > 1:
    print('B')
    
print('bye!')
```

b) The 2 statements are dependant from each other.

```python
weather = input("Enter the weather condition today ('sunny' or 'rainy') : ")

if weather == "sunny":
    print("Let's go to the sea!")
else:
    print("Better take the umbrella!")
    
```

c) The 2 statements here are mutually exclusive (if you are in scenario 1, you will not be in scenario 2).

```python
day = input('Enter a day: ')

if day == 'Monday':
    print('Mon')
elif day == 'Tuesday':
    print('Tue')
elif day == 'Wednesday':
    print('Wed')    
elif day == 'Thursday':
    print('Thu')
elif day == 'Friday':
    print('Fri')
elif day == 'Saturday':
    print('Sat')
elif day == 'Sunday':
    print('Sun')
else:
    print('You have entered an invalid day.')
    
print('bye!')
```

Note:
- You can have `if` statement without `else` statement. 
- You can have `if/elif` statements without `else` statement. The `else` statement is executed when none of the previous scenarios is true.

d) Nested `if/else` (scenario whereby there are big group scenarios and the sub-group scenarios. (Example : the insurance premium. If M or F, below that condition on age...)

```python
if a < 10:
    if a < 0:
        b = 1
    else:
        b = 2
else:
    if a < 20:
        b = 3
    else:
        b = 30
```

### One usage of tuples : when a function returns multiple values

```python
def divide(divident, divisor):
    
    quotient = divident // divisor
    remainder = divident % divisor
    
    return (quotient, remainder)
```

```python
result = divide(17, 4)
print(result)
```
