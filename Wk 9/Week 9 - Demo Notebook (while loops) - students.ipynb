### <font color = 'purple'>When do you use a while loop compared to for-loop?</font>


#### <font color = 'green'>Typically we use while loop when we want to keep prompting the user until certain conditions are fulfilled.</font>

#### <font color = 'green'>Here we want to keep prompting the user until he/she enters an even number.</font>

```python
# actions to do before the loop

# loop: actions to repeat
while not(condition to leave the loop):
    # actions to be repeated
    
          
# actions to do after the loop
          
```

```python
# actions to do before the loop
num = int(input('Please, enter an even number: '))

# loop: actions to repeat
while not(num % 2 == 0):
    print('The number is not even.')
    num = int(input('Enter a number again: '))

# actions to do after the loop
print('Thanks')
```

### <font color = 'purple'>Complex Condition</font>

#### <font color = 'green'>Prompt the user for a number and keep prompting until he/she enters a number which is a multiple of 2 and a multiple of 3.</font>

```python
num = int(input('Enter a number: '))

while not(num%2 == 0 and num%3 ==0):
    print('You didnt enter a proper number, please try again.')
    num = int(input('Enter a number: '))

print('Thank you!')
```

```python
num = int(input('Enter a number: '))

# applying De Morgan's Law
while num%2 != 0 or num%3 !=0:

    print('You didnt enter a proper number, please try again.')
    num = int(input('Enter a number: '))

print('Thank you!')
```

### Recommendations

- <font color = 'green'>(num%2 == 0 and num%3 ==0) is the condition to <b>leave</b> the loop. Just put a <b>not</b> in front.</font>

- <font color = 'green'>Make sure the <b>condition changes</b> inside the loop otherwise we may get an <b>infinite loop</b></font>

- <font color = 'green'>Sometimes if the conditions are even more complex, then it is recommended to <b>create a function</b> `is_valid()`.</font> See example below.

```python
def is_valid(my_str):

    # condition 1 - length of string must be >= 5
    if len(my_str) < 5 :
        return False
    
    # condition 2 - all characters must be digits
    if not my_str.isdigit():
        return False
    
    # condition 3 - sum of all digits must be >= 10
    a_sum = 0
    for ch in my_str:
        a_sum += int(ch)
    if a_sum < 10:
        return False
    
    return True   
```

```python
a_str = input('Please enter a string: ')

while not is_valid(a_str):
    print('You have NOT entered a valid string.')
    a_str = input('Please enter a string again: ')
    
print('Thanks!')
```

```python
# Password validation (should contain at least 10 characters and no space)

def password_is_valid(my_str):

    # condition 1 - length of password must be >= 10
    if len(my_str) < 10 :
        return False
    
    # condition 2 - password should not contain spaces
    if ' ' in my_str:
        return False
    
    return True   
```
