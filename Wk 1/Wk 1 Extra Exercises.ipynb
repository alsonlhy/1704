Q1: [**] Digits
Write a program that prompts the user for a 5-digit number, then prints the sum of the digits. (you may assume at this point of progress of learning that the input has 5 digits)
Output of the program:
Enter number: 34901
3 + 4 + 9 + 0 + 1 = 17 


```python
number = (input("Enter a 5-digit number: "))

digit_sum = int(number[0]) + int(number[1]) + int(number[2]) + int(number[3]) + int(number[4])

print(f"Your input was {number}, the sum of these digits are {number[0]} + {number[1]} + {number[2]} + {number[3]} + {number[4]} = {digit_sum}")
```

```python
number = input("Enter a 5-digit number: ")
digit_sum = 0

for i in number:
    digit_sum += int(i)

print(f"Your input was {number}, the sum of these digits are {digit_sum}")
```
